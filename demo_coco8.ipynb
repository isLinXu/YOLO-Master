# YOLO-Master COCO8 Training and Inference Demo

This notebook demonstrates how to train and run inference using the YOLO-Master model on the COCO8 dataset.

```python
import os
import warnings
from ultralytics import YOLO
from IPython.display import display, Image
import matplotlib.pyplot as plt
%matplotlib inline

warnings.filterwarnings('ignore')
```

## 1. Initialize the Model

We load the YOLO-Master Nano model configuration. 
Note: We are building the model from scratch (YAML), but you can also load pretrained weights if available.

```python
# Define the model configuration path
model_config = 'ultralytics/cfg/models/master/v0/yolo_master_n.yaml'

# Initialize the model
model = YOLO(model_config)
```

## 2. Train on COCO8

We train the model on the `coco8.yaml` dataset. This dataset is small (8 images) and perfect for quick demos and CI tests.
We'll run for a few epochs to verify the pipeline.

```python
# Train the model
results = model.train(
    data='coco128.yaml', 
    epochs=5, 
    imgsz=640, 
    project='runs/train',
    name='yolo_master_coco8',
    exist_ok=True,
    device='mps'
)
```

## 3. Validate the Model

Evaluate the model's performance on the validation set of COCO8.

```python
# Validate the model
metrics = model.val()
```

## 4. Inference and Visualization

Run inference on a sample image from the COCO8 dataset and display the results.

```python
# Run inference on a sample image
# COCO8 usually downloads to datasets/coco8 or ../datasets/coco8 depending on settings
# We'll try to find an image in the local datasets folder or download a sample if needed

import glob

# Try to find a validation image from the downloaded dataset
search_paths = [
    'datasets/coco8/images/val/*.jpg',
    '../datasets/coco8/images/val/*.jpg',
    'ultralytics/assets/bus.jpg'  # Fallback
]

image_path = None
for pattern in search_paths:
    files = glob.glob(pattern)
    if files:
        image_path = files[0]
        break

if image_path:
    print(f"Running inference on: {image_path}")
    results = model(image_path)
    
    # Display result
    for r in results:
        im_array = r.plot()  # plot a BGR numpy array of predictions
        im_rgb = im_array[..., ::-1]  # Convert BGR to RGB
        
        plt.figure(figsize=(10, 10))
        plt.imshow(im_rgb)
        plt.axis('off')
        plt.show()
else:
    print("No sample image found to run inference on.")
```
